```python
import pandas as pd
```

# standings block

```python
standings_df = pd.read_csv('lalig2_standings_2024.csv')
standings_df
```

```python
standings_df[['W', 'D', 'L']] = standings_df['W-D-L'].str.split('-', expand=True)
standings_df[['W', 'D', 'L']] = standings_df[['W', 'D', 'L']].astype(int)

standings_df = standings_df.drop(columns=['W-D-L'])

standings_df.head()
```

```python
standings_df.rename(columns={'CLUBS':'team',
                             'POSITION': 'position',
                             'MP': 'matches_played',
                             "PTS": 'points',
                             "GF": 'goals_for',
                             "GA": 'goals_against',
                             "GD": 'goals_difference',
                             "HOME": 'home_record',
                             "AWAY": 'away_record'
                             }, 
                    inplace=True) 
standings_df
```

# goalkeeping cleaning

```python
gk = pd.read_csv('laliga2_team_stats_goalkeeping_2024.csv')
gk
```

```python
gk.rename(columns={'TEAMS':'team',
                   'POSITION': 'position',
                             'GP': 'matches_played',
                             'GA': 'goals_against',
                             "S": 'shots',
                             "GD": 'goals_difference',
                             "SOG": 'shots_on_goal',
                             "SV": 'saves',
                             "CLR": "clearances",
                             "CS": 'clean_sheets'
                             }, 
                    inplace=True) 

gk.drop('clearances', axis=1, inplace=True)

gk
```

# standard cleaning

```python
standard= pd.read_csv('laliga2_team_stats_standard_2024.csv')
standard
```
