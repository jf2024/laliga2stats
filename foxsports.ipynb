```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.support.ui import Select
from selenium.webdriver.common.by import By

import pandas as pd
```

```python
website = 'https://www.foxsports.com/soccer/laliga-2/standings'
```

```python
driver = webdriver.Chrome()
driver.maximize_window()
driver.get(website)
```

# grabbing headers for later use

```python
headers_tag = driver.find_element(By.XPATH, "//thead[@class='data-header fs-11 ls-pt25']")
headers_tag
```

```python
list_headers = headers_tag.find_elements(By.TAG_NAME, "th")
list_headers
```

```python
len(list_headers)
```

```python
headers = []

for head in list_headers:
    text = head.text
    headers.append(text)

headers
```

```python
headers.insert(0, 'POSITION')
```

```python
headers
```

# first will grab standings

```python
team_rows = driver.find_elements(By.XPATH, '//tr[starts-with(@id, "tbl-row-")]')

# Now you can loop through them safely!
for row in team_rows:
    print(row.text)
    print("-" * 20)
```

# create the csv for the standings data

```python
all_team_data = []

for row in team_rows:
    row_cells = row.text.split('\n')
    row_cells = [cell.strip() for cell in row_cells if cell.strip()]
    
    all_team_data.append(row_cells)
```

```python
all_team_data[0]
```

```python
df = pd.DataFrame(all_team_data, columns=headers)
df.head()
```

## save to csv file

### can clean it up later

```python
df.to_csv("laliga2_standings.csv", index=False)
print("created standings data")
```
