```python
driver = webdriver.Chrome()
driver.maximize_window()
# Visit the website
driver.get(website)
```

```python
divs = driver.find_elements(By.XPATH,'//div[@class="css-1fip8eu-StatContainerCSS e1uxaz294"]')
divs 
```

```python
all_stat_urls = []
```

```python
for d in divs:
    boxes = d.find_elements(By.XPATH, './/div[@class="css-1wb2t24-CardCSS e1m5w55z1"]')
    print(boxes)

    for c in boxes:
        link_element = c.find_element(By.TAG_NAME, 'a')
        url = link_element.get_attribute('href')
        all_stat_urls.append(url)

        print(url)

    print("-" * 40)
```

```python
all_stat_urls
```

```python
driver.get(all_stat_urls[0])
```
